```python
# === 2. Chuyển đổi thang đo Likert về số ===
mapping = {
    "Hoàn toàn đồng ý": 5,
    "Đồng ý": 4,
    "Trung lập": 3,
    "Không đồng ý": 2,
    "Hoàn toàn không đồng ý": 1
}
data = data.replace(mapping)
```

```python
data.head()
```

```python
# === 3. Đổi tên cột theo mô hình TAM mở rộng ===
rename_dict = {}
group_prefix = {
    "CẢM NHẬN VỀ DỄ SỬ DỤNG": "PEOU",
    "CẢM NHẬN VỀ TÍNH HỮU ÍCH": "PU",
    "HỨNG THÚ KHI SỬ DỤNG": "PE",
    "ẢNH HƯỞNG XÃ HỘI": "SI",
    "NĂNG LỰC CÔNG NGHỆ CÁ NHÂN": "CSE",
    "NHẬN THỨC  RỦI RO": "PR",
    "Ý ĐỊNH SỬ DỤNG": "BI",
    "MỨC ĐỘ SỬ DỤNG": "AU"
}
```

```python
# Duyệt qua danh sách cột, gán tên mới tự động
counter = {key: 1 for key in group_prefix.values()}

for col in data.columns:
    for k, prefix in group_prefix.items():
        if k in col.upper():
            new_name = f"{prefix}{counter[prefix]}"
            rename_dict[col] = new_name
            counter[prefix] += 1

# In ra danh sách đổi tên để kiểm tra
print("\nCột đổi tên:")
for old, new in rename_dict.items():
    print(f"{old} → {new}")
```

```python
# Áp dụng đổi tên
data = data.rename(columns=rename_dict)
```

```python
data.head()
```

```python
# === 4. Giữ lại các cột liên quan đến mô hình TAM ===
cols_keep = list(rename_dict.values())  + ["AU1", "AU2", "AU3"]
data_tam = data[cols_keep].copy()

# === 5. Loại bỏ dòng trống ===
data_tam = data_tam.dropna()

# === 6. Lưu file kết quả ===
output = "tam_clean.csv"
data_tam.to_csv(output, index=False, encoding="utf-8-sig")

print(f"\nĐã lưu dữ liệu sạch tại: {output}")
print(f"Số dòng hợp lệ: {len(data_tam)}")
print(f"Số cột trong mô hình: {len(data_tam.columns)}")
```
